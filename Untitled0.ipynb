```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
import os
import shutil
import glob
from sklearn.model_selection import train_test_split

# --- 1. Define Paths ---
# PASTE THE CORRECT FOLDER NAME FROM THE `!ls` COMMAND HERE
CORRECT_BASE_FOLDER = "Multi and Hyperspectral kidney Dataset"

# Check if the base directory exists
if not os.path.exists(CORRECT_BASE_FOLDER):
    print(f"ERROR: The directory '{CORRECT_BASE_FOLDER}' was not found.")
    print("Please check the name you pasted from the '!ls' command.")
else:
    # Original data paths
    normal_dir = os.path.join(CORRECT_BASE_FOLDER, 'Multispectral_normal_images')
    tumor_dir = os.path.join(CORRECT_BASE_FOLDER, 'Multispectral_tumor_images')

    # New directories we will create
    train_dir = 'train_data'
    validation_dir = 'validation_data'
    train_normal_dir = os.path.join(train_dir, 'normal')
    train_tumor_dir = os.path.join(train_dir, 'tumor')
    validation_normal_dir = os.path.join(validation_dir, 'normal')
    validation_tumor_dir = os.path.join(validation_dir, 'tumor')

    # Create the new directories
    os.makedirs(train_normal_dir, exist_ok=True)
    os.makedirs(train_tumor_dir, exist_ok=True)
    os.makedirs(validation_normal_dir, exist_ok=True)
    os.makedirs(validation_tumor_dir, exist_ok=True)

    # --- 2. Split and Copy Files ---
    def split_and_copy(source_dir, train_dest, val_dest, split_size=0.8):
        # Using glob to find all files, robust to different image formats
        files = glob.glob(os.path.join(source_dir, '*'))

        if not files:
            print(f"Warning: No files found in {source_dir}")
            return

        train_files, val_files = train_test_split(files, train_size=split_size, random_state=42)

        for f in train_files:
            shutil.copy(f, train_dest)
        for f in val_files:
            shutil.copy(f, val_dest)
        print(f"Copied {len(train_files)} to training and {len(val_files)} to validation from {source_dir}")


    # Copy normal and tumor images
    split_and_copy(normal_dir, train_normal_dir, validation_normal_dir)
    split_and_copy(tumor_dir, train_tumor_dir, validation_tumor_dir)

    print("\nData successfully reorganized!")
```

```python
# Define image size and batch size
IMG_WIDTH, IMG_HEIGHT = 150, 150
BATCH_SIZE = 32

# Create data generators
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest',
    validation_split=0.2  # We'll use 20% of the data for validation
)

# Load the training and validation data
# Replace 'path_to_your_dataset' with the actual path to your unzipped dataset directory
train_generator = train_datagen.flow_from_directory(
    '/content/train_data',  # Update this path
    target_size=(IMG_WIDTH, IMG_HEIGHT),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training'
)

validation_generator = train_datagen.flow_from_directory(
    '/content/validation_data', # Update this path
    target_size=(IMG_WIDTH, IMG_HEIGHT),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation'
)
```

```python
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(IMG_WIDTH, IMG_HEIGHT, 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(512, activation='relu'),
    layers.Dense(3, activation='softmax')  # 3 classes: Normal, Others, Tumor
])
```

```python
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
```

```python
from tensorflow.keras import layers, models

# Define image size
IMG_WIDTH, IMG_HEIGHT = 150, 150

# Re-define the model for 2 classes
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(IMG_WIDTH, IMG_HEIGHT, 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(512, activation='relu'),
    # CORRECTED: Use 1 unit and 'sigmoid' for two-class (binary) problems
    layers.Dense(1, activation='sigmoid')
])

# CORRECTED: Compile with 'binary_crossentropy' for two-class problems
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Print the model summary to confirm
model.summary()
```
