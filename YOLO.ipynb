```python
%cd /content/Retail-Store-India-New-2-1
import time
import torch
from ultralytics import YOLO

model = YOLO("best12.pt")
img = "/content/Retail-Store-India-New-2-1/train/images/20231113_153350-2_jpg.rf.043d4cad25fb570fdf175537b8ee7f52.jpg"

# Warmup
for _ in range(10):
    _ = model(img)

# Benchmark
t0 = time.time()
for _ in range(100):
    _ = model(img)
t1 = time.time()

latency = (t1 - t0) / 100 * 1000  # ms/img
print(f"Latency: {latency:.2f} ms/img")
```

```python
from ultralytics import YOLO

# Load your model
model = YOLO("best08.pt")

# Run validation
metrics = model.val(data="/content/Retail-Store-India-New-2-1/data.yaml")

# Extract mAP values
print("mAP@50:", metrics.box.map50)       # IoU=0.50
print("mAP@50-95:", metrics.box.map)      # IoU=0.50:0.95 (COCO style)
print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
```
